# Data Wrangling

Objetivo: convertir `Outcome` en una etiqueta binaria `Class` y preparar el dataset para EDA y ML.

```python
import pandas as pd
url='https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv'
df=pd.read_csv(url)
print(df.shape)
print(df[['LaunchSite','Orbit','Outcome','Class']].head())
```

```python
print('Lanzamientos por sitio:')
print(df['LaunchSite'].value_counts())
print('Órbitas:')
print(df['Orbit'].value_counts())
print('Success rate:',df['Class'].mean())
```

### Resultado clave
El conjunto final del curso contiene 90 filas y 19 columnas; `Class=1` representa aterrizaje exitoso y `Class=0` no exitoso. La tasa global de éxito de esta etapa es 66.67%.
