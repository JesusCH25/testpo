# EDA — Data Visualization

Se estudian relaciones entre FlightNumber, PayloadMass, LaunchSite, Orbit y Class.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df=pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv')
# FlightNumber vs PayloadMass por Class
sns.scatterplot(data=df,x='FlightNumber',y='PayloadMass',hue='Class')
plt.show()
```

```python
success_rate=df.groupby('Orbit')['Class'].mean().sort_values()
print((success_rate*100).round(2))
```

### Hallazgos
1. Los primeros vuelos concentran más resultados 0.
2. VLEO presenta una tasa alta en esta muestra.
3. Algunas órbitas con 100% tienen muy pocas observaciones, por lo que no deben interpretarse aisladamente.
4. La relación FlightNumber→Class sugiere mejora operacional a través del tiempo.
