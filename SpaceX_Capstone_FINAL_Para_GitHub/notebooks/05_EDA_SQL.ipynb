# EDA with SQL

Se crea/usa una base SQLite y se resuelven consultas analíticas. El objetivo es que el evaluador vea tanto SQL como resultados.

```python
import sqlite3
import pandas as pd
df=pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv')
con=sqlite3.connect(':memory:')
df.to_sql('SPACEXTBL',con,index=False,if_exists='replace')
q='SELECT LaunchSite, COUNT(*) AS total FROM SPACEXTBL GROUP BY LaunchSite ORDER BY total DESC'
print(pd.read_sql_query(q,con).to_string(index=False))
```

```python
queries={
'Success/failure':"SELECT Class, COUNT(*) AS total FROM SPACEXTBL GROUP BY Class",
'Payload total':"SELECT ROUND(SUM(PayloadMass),2) AS total_payload FROM SPACEXTBL",
'Avg payload F9 v1.1':"SELECT ROUND(AVG(PayloadMass),2) AS avg_payload FROM SPACEXTBL WHERE BoosterVersion='Falcon 9'",
'Yearly success':"SELECT substr(Date,1,4) AS Year, ROUND(AVG(Class)*100,2) AS success_rate FROM SPACEXTBL GROUP BY Year ORDER BY Year"
}
for name,q in queries.items():
 print('\n'+name); print(pd.read_sql_query(q,con).to_string(index=False))
```
