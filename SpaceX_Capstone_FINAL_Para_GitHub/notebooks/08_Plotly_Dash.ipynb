# Plotly Dash

Dashboard para explorar landing success por sitio y rango de payload.

```python
import pandas as pd
import plotly.express as px
# El archivo se encuentra en data/ cuando se descarga el dataset del curso.
df=pd.read_csv('data/spacex_launch_dash.csv')
fig=px.pie(df,names='class',title='Landing success')
fig.show()
```
