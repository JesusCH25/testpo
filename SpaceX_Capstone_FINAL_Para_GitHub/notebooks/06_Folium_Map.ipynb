# Folium — Launch Site Locations

El mapa muestra los sitios de lanzamiento y los registros históricos.

```python
import folium
from folium.plugins import MarkerCluster
import pandas as pd
df=pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/spacex_launch_geo.csv')
m=folium.Map(location=[28.5,-95],zoom_start=4)
cluster=MarkerCluster().add_to(m)
for _,r in df.iterrows():
 folium.Marker([r['Latitude'],r['Longitude']],popup=str(r.get('Launch Site','Launch'))).add_to(cluster)
m
```

### Análisis
Los marcadores permiten explorar concentración geográfica y proximidad al litoral. El entregable `outputs/folium_launch_sites.html` conserva la versión interactiva.
