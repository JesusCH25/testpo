# Data Collection — SpaceX API

Objetivo: obtener lanzamientos de SpaceX, filtrar Falcon 9 y preparar los campos del proyecto.

```python
import requests
import pandas as pd
url='https://api.spacexdata.com/v4/launches'
r=requests.get(url,timeout=30)
r.raise_for_status()
raw=r.json()
df=pd.json_normalize(raw)
print('Registros API:', len(df))
print(df.shape)
```
